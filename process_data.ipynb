```python
import os
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
#50x50 pixels for images
```

```python
img_sizes = [100, 150, 200]


#folder locations for image files
benign_training_folder = "melanoma_cancer_dataset/train/benign/"
malignant_training_folder = "melanoma_cancer_dataset/train/malignant/"


benign_test_folder = "melanoma_cancer_dataset/test/benign/"
malignant_test_folder = "melanoma_cancer_dataset/test/malignant/"

for img_size in img_sizes:


    


    benign_training_data = []
    malignant_training_data = []

    benign_testing_data = []
    malignant_testing_data = []



    for filename in os.listdir(benign_training_folder):
        try:

            path = benign_training_folder + filename

            #converting image to color scale
            img = cv2.imread(path, cv2.IMREAD_COLOR)

            
            #resizing image
            img = cv2.resize(img, (img_size,img_size))


            # plt.imshow(img)
            # plt.show()
            # break

            # one-hot vectors
            # [1,0] = benign
            # [0,1] = melanoma

            benign_training_data.append([np.array(img), np.array([1,0])])

        except:
            pass



    for filename in os.listdir(malignant_training_folder):
        try:

            path = malignant_training_folder + filename

            #converting image to color scale
            img = cv2.imread(path, cv2.IMREAD_COLOR)

            
            #resizing image
            img = cv2.resize(img, (img_size,img_size))


            # plt.imshow(img)
            # plt.show()
            # break

            # one-hot vectors
            # [1,0] = benign
            # [0,1] = melanoma

            malignant_training_data.append([np.array(img), np.array([0,1])])

        except:
            pass




    for filename in os.listdir(benign_test_folder):
        try:

            path = benign_test_folder + filename

            #converting image to color scale
            img = cv2.imread(path, cv2.IMREAD_COLOR)

            
            #resizing image
            img = cv2.resize(img, (img_size,img_size))


            # plt.imshow(img)
            # plt.show()
            # break

            # one-hot vectors
            # [1,0] = benign
            # [0,1] = melanoma

            benign_testing_data.append([np.array(img), np.array([1,0])])

        except:
            pass







    for filename in os.listdir(malignant_test_folder):
        try:

            path = malignant_test_folder + filename

            #converting image to color scale
            img = cv2.imread(path, cv2.IMREAD_COLOR)

            
            #resizing image
            img = cv2.resize(img, (img_size,img_size))


            # plt.imshow(img)
            # plt.show()
            # break

            # one-hot vectors
            # [1,0] = benign
            # [0,1] = melanoma

            malignant_testing_data.append([np.array(img), np.array([0,1])])

        except:
            pass





    benign_training_data = benign_training_data[0:len(malignant_training_data)]





    print()
    print()
    print(f"Benign training count {img_size}x{img_size}: {len(benign_training_data)}")
    print(f"Malignant training count {img_size}x{img_size}: {len(malignant_training_data)}")
    print()
    print(f"Benign testing count {img_size}x{img_size}: {len(benign_testing_data)}")
    print(f"Malignant testing count {img_size}x{img_size}: {len(malignant_testing_data)}")




    #combine training data
    training_data = benign_training_data + malignant_training_data
    #combine testing data
    testing_data = benign_testing_data + malignant_testing_data

    #randomize data
    np.random.shuffle(training_data)
    np.random.shuffle(testing_data)




    #save as numpy object
    np.save(f"melanoma_training_data_{img_size}.npy", np.array(training_data, dtype=object), allow_pickle=True)
    np.save(f"melanoma_testing_data_{img_size}.npy", np.array(testing_data, dtype=object), allow_pickle=True)


```
